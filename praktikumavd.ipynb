```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style('whitegrid')
df = pd.read_csv('penjualan_nusantaramart.csv')

print("=== 5 BARIS PERTAMA ===")
print(df.head())

print("\n=== INFO DATASET ===")
print(df.info())

print("\n=== RINGKASAN STATISTIK ===")
print(df.describe())

# Rata-rata total pembelian per kategori produk
rata_kategori = df.groupby('kategori_produk')['total'].mean()
print("\n=== RATA-RATA TOTAL PER KATEGORI ===")
print(rata_kategori.sort_values(ascending=False))
```

#Praktikum 1
1. Kategori produk yang punya rata-rata nilai transaksi (total) paling tinggi adalah Elektronik dengan rata-rata sekitar Rp 363.541.
2. Tidak ada kolom yang datanya kosong. Cara mengeceknya dari output df.info() tinggal lihat bagian Non-Null Count di setiap kolom. Semuanya tertulis 220 non-null, sama seperti jumlah total baris datanya (220 entries), berarti datanya terisi lengkap.
3. Nilai mean kolom total adalah Rp 238.486, sedangkan mediannya Rp 125.000. Nilai mean jauh lebih besar dari median karena ada beberapa transaksi yang nilainya besar banget (outlier), jadi nilai rata-ratanya ikut ketarik ke atas. Karena itu, nilai median lebih cocok buat menggambarkan transaksi normal pembeli pada umumnya.

```python
total_kategori = df.groupby('kategori_produk')['total'].sum().sort_values(ascending=False)
print(total_kategori)

plt.figure(figsize=(8, 5))
plt.title('Total Penjualan per Kategori Produk')
sns.barplot(x=total_kategori.values, y=total_kategori.index, hue=total_kategori.index, palette='Blues_r', legend=False)
plt.xlabel('Total Penjualan (Rp)')
plt.ylabel('Kategori Produk')
plt.tight_layout()
plt.show()
```

#Praktikum 2
1. Kategori yang menyumbang penjualan tertinggi adalah Fashion (Rp 17.695.000) dan yang paling rendah adalah Makanan & Minuman (Rp 3.029.000).
2. Rekomendasi untuk manajer Nusantara Mart: stok barang untuk kategori Fashion dan Elektronik harus selalu aman karena dua kategori ini yang paling laku dan menyumbang pemasukan paling besar. Sementara untuk kategori Makanan & Minuman bisa dikasih promo paket hemat, diskon, atau ditaruh di halaman depan aplikasi supaya lebih banyak yang beli.

```python
df['tanggal'] = pd.to_datetime(df['tanggal'])
df['bulan'] = df['tanggal'].dt.to_period('M')
tren_bulanan = df.groupby('bulan')['total'].sum()
print(tren_bulanan)

plt.figure(figsize=(8, 5))
plt.plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o', color='crimson')
plt.title('Tren Penjualan Bulanan')
plt.xlabel('Bulan')
plt.ylabel('Total Penjualan (Rp)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

#Praktikum 3
1. Bulan dengan penjualan tertinggi adalah bulan Mei 2026 (2026-05) dengan total Rp 13.585.000. Kemungkinan penyebabnya karena ada promo tanggal kembar (5.5), promo gajian, atau ada pelanggan yang beli dalam jumlah banyak di bulan itu.
2. Tren penjualannya cenderung fluktuatif. Awalnya selalu naik dari bulan Januari sampai Mei, tapi pas masuk bulan Juni turun drastis jadi Rp 6.992.000. Penurunan ini kemungkinan karena promo besar di bulan Mei sudah selesai atau data transaksi bulan Juni belum tercatat satu bulan penuh.

```python
plt.figure(figsize=(7, 5))
sns.scatterplot(data=df, x='jumlah', y='total', hue='kategori_produk')
plt.title('Hubungan Jumlah Pembelian dengan Total Nilai Transaksi')
plt.xlabel('Jumlah Barang')
plt.ylabel('Total (Rp)')
plt.tight_layout()
plt.show()
```

#Praktikum 4
1. Terlihat pola hubungan yang positif antara jumlah barang dan total nilai transaksi. Artinya, semakin banyak jumlah barang yang dibeli, total harganya juga semakin besar.
2. Ada beberapa titik data yang jauh dari pola umum (outlier) di bagian kanan grafik, yaitu pembelian sebanyak 25 dan 40 barang. Dari warnanya, outlier ini ada di kategori Elektronik, Olahraga, dan Kesehatan.

```python
# Histogram distribusi rating pelanggan
plt.figure(figsize=(7, 5))
sns.histplot(df['rating_pelanggan'], bins=5, color='seagreen')
plt.title('Distribusi Rating Pelanggan')
plt.xlabel('Rating')
plt.ylabel('Frekuensi')
plt.tight_layout()
plt.show()

# Boxplot jumlah pembelian per kategori (deteksi outlier)
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x='kategori_produk', y='jumlah')
plt.title('Sebaran Jumlah Pembelian per Kategori Produk')
plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

#Praktikum 5
1. Distribusi rating pelanggan cenderung tinggi karena sebagian besar pembeli memberikan rating 4 dan 5 (rata-ratanya 4,15) dan tidak ada yang memberi rating 1. Ini tandanya pelanggan sudah cukup puas dengan produknya.
2. Kategori produk yang punya titik-titik di luar kotak boxplot (outlier) adalah Elektronik, Olahraga, Kesehatan, dan Fashion. Kemungkinan penyebabnya karena ada pembelian dalam jumlah banyak sekaligus, misalnya dibeli oleh reseller untuk dijual lagi atau buat kebutuhan acara dan kantor.

```python
fig, axes = plt.subplots(2, 2, figsize=(13, 9))

sns.barplot(x=total_kategori.values, y=total_kategori.index, hue=total_kategori.index, palette='Blues_r', legend=False, ax=axes[0,0])
axes[0,0].set_title('Total Penjualan per Kategori')

axes[0,1].plot(tren_bulanan.index.astype(str), tren_bulanan.values, marker='o', color='crimson')
axes[0,1].set_title('Tren Penjualan Bulanan')
axes[0,1].tick_params(axis='x', rotation=45)

sns.histplot(df['rating_pelanggan'], bins=5, color='seagreen', ax=axes[1,0])
axes[1,0].set_title('Distribusi Rating Pelanggan')

sns.boxplot(data=df, x='kategori_produk', y='jumlah', ax=axes[1,1])
axes[1,1].set_title('Sebaran Jumlah per Kategori')
axes[1,1].tick_params(axis='x', rotation=20)

plt.tight_layout()
plt.show()
```
